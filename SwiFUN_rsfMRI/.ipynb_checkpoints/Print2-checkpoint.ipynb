```python
# ---- 读取并绘制 rollout 曲线（健壮版）----
import re, numpy as np, pandas as pd, matplotlib.pyplot as plt
from wandb import Api

ENTITY  = "mengchuz-carnegie-mellon-university"
PROJECT = "ds000030_rsfMRI_alldata_6context_6pred_10kmax"
RUN_ID  = "56an1qq8"

api = Api()

# 取到这个 run（兼容不同路径写法）
def get_run(api, entity, project, run_id):
    try:
        return api.run(f"{entity}/{project}/{run_id}")
    except Exception:
        try:
            return api.run(f"{entity}/{project}/runs/{run_id}")
        except Exception:
            for r in api.runs(f"{entity}/{project}"):
                if r.id == run_id or r.name == run_id:
                    return r
            raise RuntimeError(f"找不到 run {run_id}")

run = get_run(api, ENTITY, PROJECT, RUN_ID)
print("找到 run:", run.id, "| URL:", run.url)

# 历史日志
rows = [row for row in run.scan_history()]  # 比 history(pandas=True) 更省内存
hist = pd.DataFrame(rows)
print("历史条数:", len(hist))

def last_non_nan(series: pd.Series) -> float:
    s = pd.to_numeric(series, errors="coerce").dropna()
    return float(s.iloc[-1]) if not s.empty else np.nan

# 收集所有包含 @k 的列
def collect_k_cols(hist: pd.DataFrame, prefix: str):
    bucket = {}  # k -> [cols...]
    for c in hist.columns:
        if isinstance(c, str) and c.startswith(prefix):
            m = re.search(r"@(\d+)", c)     # 只要出现 @数字 就算
            if not m:
                continue
            k = int(m.group(1))
            bucket.setdefault(k, []).append(c)
    return bucket

mse_map = collect_k_cols(hist, "valid_mse@")
r_map   = collect_k_cols(hist, "valid_r@")
ks = sorted(set(mse_map.keys()) | set(r_map.keys()))

# 辅助：同一个 k 可能有多个候选列，取“最后一个非 NaN 值”
def value_for_k(col_list):
    vals = [last_non_nan(hist.get(c, pd.Series(dtype=float))) for c in col_list]
    vals = [v for v in vals if np.isfinite(v)]
    return vals[-1] if vals else np.nan

mse_vals = np.array([value_for_k(mse_map.get(k, [])) for k in ks], dtype=float)
r_vals   = np.array([value_for_k(r_map.get(k,   [])) for k in ks], dtype=float)

# 同时读取 baseline（如果没记录就会是 NaN）
def take(key): 
    return last_non_nan(hist.get(key, pd.Series(dtype=float)))

mse_persist = take("valid_mse_baseline_persist")
mse_ctxmean = take("valid_mse_baseline_ctxmean")
r_persist   = take("valid_r_time_baseline_persist")
r_ctxmean   = take("valid_r_time_baseline_ctxmean")

print("找到的 ks:", ks)
print("MSE@k:", mse_vals)
print("r@k  :", r_vals)

# 可视化
plt.figure(figsize=(6,4))
plt.plot(ks, mse_vals, marker="o", label="model MSE@k")
if np.isfinite(mse_persist): plt.axhline(mse_persist, ls="--", label="baseline: persistence")
if np.isfinite(mse_ctxmean): plt.axhline(mse_ctxmean, ls=":",  label="baseline: ctx-mean")
plt.xlabel("k (steps ahead)"); plt.ylabel("MSE@k"); plt.title(f"Rollout MSE@k — {RUN_ID}")
plt.legend(); plt.tight_layout(); plt.show()

plt.figure(figsize=(6,4))
plt.plot(ks, r_vals, marker="o", label="model r@k")
if np.isfinite(r_persist): plt.axhline(r_persist, ls="--", label="baseline: persistence")
if np.isfinite(r_ctxmean): plt.axhline(r_ctxmean, ls=":",  label="baseline: ctx-mean")
plt.xlabel("k (steps ahead)"); plt.ylabel("Pearson r@k"); plt.title(f"Rollout r@k — {RUN_ID}")
plt.legend(); plt.tight_layout(); plt.show()
```

```python
# --- Plot rollout from an already-finished W&B run (no retrain needed) ---
import re, math, numpy as np, pandas as pd, matplotlib.pyplot as plt
from wandb import Api

ENTITY  = "mengchuz-carnegie-mellon-university"
PROJECT = "ds000030_rsfMRI_alldata_6context_6pred_10kmax"
RUN_ID  = "56an1qq8"

api = Api()

def get_run(api, entity, project, run_id):
    for path in (f"{entity}/{project}/{run_id}", f"{entity}/{project}/runs/{run_id}"):
        try:
            return api.run(path)
        except Exception:
            pass
    # fallback: list & match id/name
    for r in api.runs(f"{entity}/{project}"):
        if r.id == run_id or r.name == run_id:
            return r
    raise RuntimeError(f"找不到 run {run_id}")

run = get_run(api, ENTITY, PROJECT, RUN_ID)
print("找到 run:", run.id, "| URL:", run.url)

# ---- read history (streaming) ----
rows = [row for row in run.scan_history()]
hist = pd.DataFrame(rows)
print("history 条数:", len(hist))

def last_non_nan_col(df, col):
    s = pd.to_numeric(df[col], errors="coerce").dropna()
    return float(s.iloc[-1]) if not s.empty else np.nan

# collect k columns (works even with suffixes)
def collect_k_map(df, kind):  # kind in {"mse","r"}
    pat = rf"valid_{kind}@(\d+)"
    bucket = {}
    for c in df.columns:
        if isinstance(c, str):
            m = re.search(pat, c)
            if m:
                k = int(m.group(1))
                bucket.setdefault(k, []).append(c)
    return bucket

mse_map = collect_k_map(hist, "mse")
r_map   = collect_k_map(hist, "r")
ks = sorted(set(mse_map) | set(r_map))
if not ks:
    raise RuntimeError("历史中没发现 'valid_mse@k' / 'valid_r@k'。")

def value_for_k(cols):
    vals = []
    for c in cols:
        v = last_non_nan_col(hist, c)
        if np.isfinite(v):
            vals.append(v)
    return vals[-1] if vals else np.nan

mse_vals = np.array([value_for_k(mse_map.get(k, [])) for k in ks], float)
r_vals   = np.array([value_for_k(r_map.get(k,   [])) for k in ks], float)

# ---- find baselines from history or summary (substring match) ----
def fetch_baseline(substr):
    # try history first
    cols = [c for c in hist.columns if isinstance(c, str) and substr in c]
    for c in cols[::-1]:
        v = last_non_nan_col(hist, c)
        if np.isfinite(v):
            return v, c
    # fallback to summary
    summ = dict(run.summary)
    cand = [k for k in summ if isinstance(k, str) and substr in k]
    for k in sorted(cand)[::-1]:
        try:
            v = float(summ[k])
            if np.isfinite(v): return v, f"[summary]{k}"
        except Exception:
            pass
    return np.nan, None

mse_persist, key_mp = fetch_baseline("mse_baseline_persist")
mse_ctxmean, key_mm = fetch_baseline("mse_baseline_ctxmean")
r_persist,   key_rp = fetch_baseline("r_time_baseline_persist")
r_ctxmean,   key_rm = fetch_baseline("r_time_baseline_ctxmean")

print("[baseline 捕获]")
print(" MSE persist :", key_mp, mse_persist)
print(" MSE ctxmean :", key_mm, mse_ctxmean)
print(" r   persist :", key_rp, r_persist)
print(" r   ctxmean :", key_rm, r_ctxmean)

# ---- pred_horizon for the vertical line (auto or manual) ----
cfg = dict(run.config) if hasattr(run, "config") else {}
H = int(cfg.get("pred_horizon", 0)) or None  # 手动指定可写：H = 6

# ---- Fig 1/2: 1-step rollout curves with baselines ----
plt.figure(figsize=(6,4))
plt.plot(ks, mse_vals, marker="o", label="model MSE@k")
if np.isfinite(mse_persist): plt.axhline(mse_persist, ls="--", label="baseline: persistence")
if np.isfinite(mse_ctxmean): plt.axhline(mse_ctxmean, ls=":",  label="baseline: ctx-mean")
if H: plt.axvline(H, ls="--", label="pred_horizon")
plt.xlabel("k (steps ahead)"); plt.ylabel("MSE@k"); plt.title(f"Rollout MSE@k — {RUN_ID}")
plt.legend(); plt.tight_layout(); plt.show()

plt.figure(figsize=(6,4))
plt.plot(ks, r_vals, marker="o", label="model r@k")
if np.isfinite(r_persist): plt.axhline(r_persist, ls="--", label="baseline: persistence")
if np.isfinite(r_ctxmean): plt.axhline(r_ctxmean, ls=":",  label="baseline: ctx-mean")
if H: plt.axvline(H, ls="--", label="pred_horizon")
plt.xlabel("k (steps ahead)"); plt.ylabel("Pearson r@k"); plt.title(f"Rollout r@k — {RUN_ID}")
plt.legend(); plt.tight_layout(); plt.show()

# ---- optional: block-of-H aggregation (allow partial blocks) ----
def block_means(ks, vals, H, partial_ok=True):
    if not H or H < 2: return [], np.array([])
    max_k = max(ks)
    n_blocks = math.ceil(max_k / H) if partial_ok else (max_k // H)
    ids, out = [], []
    for b in range(1, n_blocks+1):
        members = [vals[i] for i,k in enumerate(ks) if ((b-1)*H < k <= min(b*H, max_k))]
        if not members: continue
        out.append(np.nanmean(members))
        ids.append(b)
    return ids, np.array(out, float)

if H and H >= 2:
    bid_mse, mse_block = block_means(ks, mse_vals, H, partial_ok=True)
    bid_r,   r_block   = block_means(ks, r_vals,   H, partial_ok=True)

    if bid_mse:
        plt.figure(figsize=(6,4))
        plt.plot(bid_mse, mse_block, marker="o", label=f"block-avg MSE (H={H})")
        if np.isfinite(mse_persist): plt.axhline(mse_persist, ls="--", label="baseline: persistence")
        if np.isfinite(mse_ctxmean): plt.axhline(mse_ctxmean, ls=":",  label="baseline: ctx-mean")
        plt.xlabel("block index"); plt.ylabel("MSE (avg over H)"); plt.title(f"Block-of-{H} rollout — {RUN_ID}")
        plt.legend(); plt.tight_layout(); plt.show()

    if bid_r:
        plt.figure(figsize=(6,4))
        plt.plot(bid_r, r_block, marker="o", label=f"block-avg r (H={H})")
        if np.isfinite(r_persist): plt.axhline(r_persist, ls="--", label="baseline: persistence")
        if np.isfinite(r_ctxmean): plt.axhline(r_ctxmean, ls=":",  label="baseline: ctx-mean")
        plt.xlabel("block index"); plt.ylabel("r (avg over H)"); plt.title(f"Block-of-{H} rollout — {RUN_ID}")
        plt.legend(); plt.tight_layout(); plt.show()
```

```python
# --- Plot rollout from an already-finished W&B run (no retrain needed) ---
import re, math, numpy as np, pandas as pd, matplotlib.pyplot as plt
from wandb import Api

def get_run(api, entity, project, run_id):
    for path in (f"{entity}/{project}/{run_id}", f"{entity}/{project}/runs/{run_id}"):
        try:
            return api.run(path)
        except Exception:
            pass
    for r in api.runs(f"{entity}/{project}"):
        if r.id == run_id or r.name == run_id:
            return r
    raise RuntimeError(f"找不到 run {run_id}")

run = get_run(api, ENTITY, PROJECT, RUN_ID)
print("找到 run:", run.id, "| URL:", run.url)

# ---- read history (streaming) ----
rows = [row for row in run.scan_history()]
hist = pd.DataFrame(rows)
print("history 条数:", len(hist))

def last_non_nan_series(s: pd.Series):
    s = pd.to_numeric(s, errors="coerce")
    idx = np.flatnonzero(s.notna().to_numpy())
    return (float(s.iloc[idx[-1]]), int(idx[-1])) if idx.size else (np.nan, -1)

def last_non_nan_col(df, col):
    if col not in df.columns: return np.nan
    v, _ = last_non_nan_series(df[col])
    return v

# collect k columns (works even with suffixes)
def collect_k_map(df, kind):  # kind in {"mse","r"}
    pat = rf"valid_{kind}@(\d+)"
    bucket = {}
    for c in df.columns:
        if isinstance(c, str):
            m = re.search(pat, c)
            if m:
                k = int(m.group(1))
                bucket.setdefault(k, []).append(c)
    return bucket

mse_map = collect_k_map(hist, "mse")
r_map   = collect_k_map(hist, "r")
ks = sorted(set(mse_map) | set(r_map))
if not ks:
    raise RuntimeError("历史中没发现 'valid_mse@k' / 'valid_r@k'。")

def value_for_k(cols):
    best_val, best_pos = np.nan, -1
    for c in cols:
        v, pos = last_non_nan_series(hist[c])
        if np.isfinite(v) and pos > best_pos:
            best_val, best_pos = v, pos
    return best_val

mse_vals = np.array([value_for_k(mse_map.get(k, [])) for k in ks], float)
r_vals   = np.array([value_for_k(r_map.get(k,   [])) for k in ks], float)

# ---- baseline picker: prefer /dataloader_idx_{i} -> _epoch -> raw ----
def pick_baseline(hist: pd.DataFrame, stem: str, prefer_idx=0, modes=("valid","test")):
    """
    stem: 'mse_baseline_persist' / 'mse_baseline_ctxmean'
          'r_time_baseline_persist' / 'r_time_baseline_ctxmean'
    return: value, chosen_key
    """
    tiers = [
        f"dataloader_idx_{prefer_idx}",    # 最高优先
        "_epoch",
        "",                                # 回退
    ]
    best_key, best_val, best_pos = None, np.nan, -1
    for mode in modes:
        for tier in tiers:
            # 任意位置包含 tier 即算（兼容 '/dataloader_idx_' 或 '_dataloader_idx_')
            cand = [c for c in hist.columns
                    if isinstance(c, str) and (f"{mode}_{stem}" in c) and (tier in c)]
            for c in cand:
                v, pos = last_non_nan_series(hist[c])
                if np.isfinite(v) and pos > best_pos:
                    best_key, best_val, best_pos = c, v, pos
            if best_key is not None:
                break
        if best_key is not None:
            break
    return best_val, best_key

mse_persist, key_mp = pick_baseline(hist, "mse_baseline_persist", prefer_idx=PREFERRED_DLI)
mse_ctxmean, key_mm = pick_baseline(hist, "mse_baseline_ctxmean", prefer_idx=PREFERRED_DLI)
r_persist,   key_rp = pick_baseline(hist, "r_time_baseline_persist", prefer_idx=PREFERRED_DLI)
r_ctxmean,   key_rm = pick_baseline(hist, "r_time_baseline_ctxmean", prefer_idx=PREFERRED_DLI)

fmt = lambda v: f"{v:.6g}" if np.isfinite(v) else "NaN"
print("[baseline 选取]")
print(" MSE persist :", fmt(mse_persist), "<-", key_mp)
print(" MSE ctxmean :", fmt(mse_ctxmean), "<-", key_mm)
print(" r   persist :", fmt(r_persist),   "<-", key_rp)
print(" r   ctxmean :", fmt(r_ctxmean),   "<-", key_rm)

# ---- pred_horizon for the vertical line ----
cfg = dict(run.config) if hasattr(run, "config") else {}
H = int(cfg.get("pred_horizon", 0)) or None

# ---- Fig 1/2: rollout curves with colored baselines ----
plt.figure(figsize=(6,4))
plt.plot(ks, mse_vals, marker="o", lw=2, color="tab:blue", label="model MSE@k")
if np.isfinite(mse_persist): plt.axhline(mse_persist, ls="--", lw=2, color="tab:orange", label="baseline: persistence")
if np.isfinite(mse_ctxmean): plt.axhline(mse_ctxmean, ls="-.", lw=2, color="tab:green",  label="baseline: ctx-mean")
if H: plt.axvline(H, ls="--", lw=1.5, color="tab:purple", label="pred_horizon")
plt.xlabel("k (steps ahead)"); plt.ylabel("MSE@k"); plt.title(f"Rollout MSE@k — {RUN_ID}")
plt.legend(); plt.tight_layout(); plt.show()

plt.figure(figsize=(6,4))
plt.plot(ks, r_vals, marker="o", lw=2, color="tab:blue", label="model r@k")
if np.isfinite(r_persist): plt.axhline(r_persist, ls="--", lw=2, color="tab:orange", label="baseline: persistence")
if np.isfinite(r_ctxmean): plt.axhline(r_ctxmean, ls="-.", lw=2, color="tab:green",  label="baseline: ctx-mean")
if H: plt.axvline(H, ls="--", lw=1.5, color="tab:purple", label="pred_horizon")
plt.xlabel("k (steps ahead)"); plt.ylabel("Pearson r@k"); plt.title(f"Rollout r@k — {RUN_ID}")
plt.legend(); plt.tight_layout(); plt.show()

# ---- optional: block-of-H aggregation (allow partial blocks) ----
def block_means(ks, vals, H, partial_ok=True):
    if not H or H < 2: return [], np.array([])
    max_k = max(ks)
    n_blocks = math.ceil(max_k / H) if partial_ok else (max_k // H)
    ids, out = [], []
    for b in range(1, n_blocks+1):
        members = [vals[i] for i,k in enumerate(ks) if ((b-1)*H < k <= min(b*H, max_k))]
        if not members: continue
        out.append(np.nanmean(members))
        ids.append(b)
    return ids, np.array(out, float)

if H and H >= 2:
    bid_mse, mse_block = block_means(ks, mse_vals, H, partial_ok=True)
    bid_r,   r_block   = block_means(ks, r_vals,   H, partial_ok=True)

    if bid_mse:
        plt.figure(figsize=(6,4))
        plt.plot(bid_mse, mse_block, marker="o", lw=2, color="tab:blue", label=f"block-avg MSE (H={H})")
        if np.isfinite(mse_persist): plt.axhline(mse_persist, ls="--", lw=2, color="tab:orange", label="baseline: persistence")
        if np.isfinite(mse_ctxmean): plt.axhline(mse_ctxmean, ls="-.",  lw=2, color="tab:green",  label="baseline: ctx-mean")
        plt.xlabel("block index"); plt.ylabel("MSE (avg over H)"); plt.title(f"Block-of-{H} rollout — {RUN_ID}")
        plt.legend(); plt.tight_layout(); plt.show()

    if bid_r:
        plt.figure(figsize=(6,4))
        plt.plot(bid_r, r_block, marker="o", lw=2, color="tab:blue", label=f"block-avg r (H={H})")
        if np.isfinite(r_persist): plt.axhline(r_persist, ls="--", lw=2, color="tab:orange", label="baseline: persistence")
        if np.isfinite(r_ctxmean): plt.axhline(r_ctxmean, ls="-.",  lw=2, color="tab:green",  label="baseline: ctx-mean")
        plt.xlabel("block index"); plt.ylabel("r (avg over H)"); plt.title(f"Block-of-{H} rollout — {RUN_ID}")
        plt.legend(); plt.tight_layout(); plt.show()
```

```python
# === rfMRI_next predictions 可视化 Notebook ===
# 使用方法：修改 PRED_DIR 为你的 predictions 子目录（如 "./predictions/56an1qq8"）

import os, gzip, pickle, numpy as np, matplotlib.pyplot as plt

# ====== 1) 路径设置 ======
PRED_DIR = "./output/ds000030_rsfMRI_alldata_6/predictions/56an1qq8"   # <<< 改成你实际的 predictions 路径

FILES = {
    "subjects_best": os.path.join(PRED_DIR, "test_subjects_best.pkl.gz"),
    "pred_best":     os.path.join(PRED_DIR, "predicted_map_best.pkl.gz"),
    "tgt_best":      os.path.join(PRED_DIR, "target_map_best.pkl.gz"),
    "subjects_last": os.path.join(PRED_DIR, "test_subjects_last.pkl.gz"),
    "pred_last":     os.path.join(PRED_DIR, "predicted_map_last.pkl.gz"),
    "tgt_last":      os.path.join(PRED_DIR, "target_map_last.pkl.gz"),
}

# ====== 2) 工具函数 ======
def load_gz_pickle(path):
    with gzip.open(path, "rb") as f:
        return pickle.load(f)

def infer_T_horizon(flat_length, spatial=(96,96,96), fallback=None):
    V = np.prod(spatial)
    if flat_length % V == 0:
        return flat_length // V
    if fallback is not None:
        return int(fallback)
    raise ValueError(f"无法推断 T_horizon: flat_len={flat_length}, V={V}")

def flat_to_4d(arr_1d, T, spatial=(96,96,96)):
    V = np.prod(spatial)
    return arr_1d.reshape(spatial + (T,))

def r_time_mean(pred4d, tgt4d):
    V = np.prod(pred4d.shape[:-1])
    P = pred4d.reshape(V, pred4d.shape[-1])
    T = tgt4d.reshape(V, tgt4d.shape[-1])
    Pz = (P - P.mean(-1, keepdims=True)) / (P.std(-1, keepdims=True) + 1e-6)
    Tz = (T - T.mean(-1, keepdims=True)) / (T.std(-1, keepdims=True) + 1e-6)
    return float((Pz*Tz).mean())

def mse(pred4d, tgt4d):
    return float(np.mean((pred4d - tgt4d)**2))

def plot_axial_time_grid(pred4d, tgt4d, subj, out_png=None, z_slice=None):
    T = pred4d.shape[-1]
    if z_slice is None:
        z_slice = pred4d.shape[2] // 2
    vmin, vmax = min(pred4d.min(), tgt4d.min()), max(pred4d.max(), tgt4d.max())

    fig, axes = plt.subplots(2, T, figsize=(2.3*T, 4.6), constrained_layout=True)
    for t in range(T):
        axes[0, t].imshow(pred4d[:, :, z_slice, t].T, origin="lower", vmin=vmin, vmax=vmax)
        axes[0, t].set_axis_off()
        axes[0, t].set_title(f"t={t+1}")
        axes[1, t].imshow(tgt4d[:, :, z_slice, t].T, origin="lower", vmin=vmin, vmax=vmax)
        axes[1, t].set_axis_off()
    axes[0,0].set_ylabel("PRED", rotation=0, labelpad=20)
    axes[1,0].set_ylabel("GT", rotation=0, labelpad=20)
    fig.suptitle(f"Subject={subj}", fontsize=12)
    if out_png:
        fig.savefig(out_png, dpi=150)
        print(f"[SAVED] {out_png}")
    else:
        plt.show()
    plt.close(fig)

# ====== 3) 读取 best 结果 ======
subjects = load_gz_pickle(FILES["subjects_best"])
pred     = np.asarray(load_gz_pickle(FILES["pred_best"]))
tgt      = np.asarray(load_gz_pickle(FILES["tgt_best"]))

N, flat_len = pred.shape
T_HZ = infer_T_horizon(flat_len, spatial=(96,96,96))
print(f"[INFO] N={N}, flat_len={flat_len}, 推断 T_horizon={T_HZ}")

# ====== 4) 计算 MSE/r_time_mean，挑出 best/median/worst 三个 subject 可视化 ======
metrics = []
for i in range(N):
    p4, g4 = flat_to_4d(pred[i], T=T_HZ), flat_to_4d(tgt[i], T=T_HZ)
    metrics.append((i, mse(p4, g4), r_time_mean(p4, g4)))

metrics = sorted(metrics, key=lambda x: x[1])  # 按 MSE 升序
idx_best, idx_med, idx_worst = metrics[0][0], metrics[len(metrics)//2][0], metrics[-1][0]

for tag, idx in [("best", idx_best), ("median", idx_med), ("worst", idx_worst)]:
    p4, g4 = flat_to_4d(pred[idx], T=T_HZ), flat_to_4d(tgt[idx], T=T_HZ)
    m, r = mse(p4, g4), r_time_mean(p4, g4)
    print(f"{tag.upper()} subj={subjects[idx]}  MSE={m:.4g}, r_time={r:.4f}")
    plot_axial_time_grid(p4, g4, subjects[idx])
```

```python
# === rfMRI_next predictions viewer (mask-aware, auto-slice, diff view) ===
# 1) 修改 PRED_DIR 为你的 predictions 子目录（例："./predictions/56an1qq8"）
# 2) 运行即可；默认读取 *best*，也可切到 *last*

import os, gzip, pickle, numpy as np, matplotlib.pyplot as plt

# ---------------- Paths ----------------
PRED_DIR = "./output/ds000030_rsfMRI_alldata_6/predictions/56an1qq8"      # <<< 改成你的目录
USE_TAG  = "best"                        # "best" 或 "last"

FILES = {
    "subjects": os.path.join(PRED_DIR, f"test_subjects_{USE_TAG}.pkl.gz"),
    "pred":     os.path.join(PRED_DIR, f"predicted_map_{USE_TAG}.pkl.gz"),
    "tgt":      os.path.join(PRED_DIR, f"target_map_{USE_TAG}.pkl.gz"),
}

# ---------------- Utils ----------------
def load_gz_pickle(path):
    with gzip.open(path, "rb") as f:
        return pickle.load(f)

def infer_T_horizon(flat_length, spatial=(96,96,96)):
    V = np.prod(spatial)
    if flat_length % V != 0:
        raise ValueError(f"Cannot infer T: flat_len={flat_length}, V={V}")
    return flat_length // V

def to_4d(arr_flat, T, spatial=(96,96,96)):
    return np.asarray(arr_flat).reshape(spatial + (T,))

def brain_mask_from_target(tgt4d, thr=0.0):
    """在任一时间帧非零（或 |x|>thr）的体素作为脑内掩膜"""
    return (np.abs(tgt4d) > thr).any(axis=-1)

def pick_best_z(mask3d):
    """选择脑内体素最多的 z 切片"""
    # mask3d shape: [X,Y,Z]
    z_counts = mask3d.sum(axis=(0,1))
    return int(np.argmax(z_counts))

def r_time_mean(pred4d, tgt4d, mask3d):
    """在 mask 内计算 mean Pearson (沿时间)"""
    P = pred4d[mask3d].reshape(-1, pred4d.shape[-1])
    T = tgt4d[mask3d].reshape(-1, tgt4d.shape[-1])
    Pz = (P - P.mean(-1, keepdims=True)) / (P.std(-1, keepdims=True) + 1e-6)
    Tz = (T - T.mean(-1, keepdims=True)) / (T.std(-1, keepdims=True) + 1e-6)
    return float((Pz * Tz).mean())

def mse_masked(pred4d, tgt4d, mask3d):
    D = (pred4d - tgt4d)[mask3d]
    return float(np.mean(D**2))

def robust_limits(a, q=0.5):
    """更稳定的色标：用分位数避免极端值(q=0.5% 和 99.5%)"""
    lo, hi = np.percentile(a, [q, 100-q])
    if lo == hi:  # 退化处理
        lo, hi = float(a.min()), float(a.max())
    if lo == hi:
        hi = lo + 1e-6
    return lo, hi

def plot_subject_grid(pred4d, tgt4d, mask3d, subj_label, save_png=None):
    """三行：PRED / GT / DIFF；每列一个时间帧；自动选择信息量最大的 z-slice"""
    T = pred4d.shape[-1]
    z = pick_best_z(mask3d)
    diff = pred4d - tgt4d

    # 色标：pred/gt 共享一个范围，diff 单独一个对称范围
    vmin, vmax = robust_limits(np.concatenate([pred4d[...,z,:].ravel(), tgt4d[...,z,:].ravel()]))
    dmax = np.max(np.abs(diff[...,z,:])); dmin = -dmax if dmax>0 else -1e-6

    fig, axes = plt.subplots(3, T, figsize=(2.4*T, 7.2), constrained_layout=True)
    rows = [("PRED", pred4d), ("GT", tgt4d), ("DIFF", diff)]
    for r, (title, vol) in enumerate(rows):
        for t in range(T):
            ax = axes[r, t]
            sl = vol[:, :, z, t].T  # axial view
            if title == "DIFF":
                im = ax.imshow(sl, origin="lower", vmin=dmin, vmax=-dmin, cmap="bwr")
            else:
                im = ax.imshow(sl, origin="lower", vmin=vmin, vmax=vmax)
            ax.set_axis_off()
            if r == 0:
                ax.set_title(f"t={t+1}")
        axes[r,0].set_ylabel(title, rotation=0, labelpad=20)

    fig.suptitle(f"Subject={subj_label} (z={z})", fontsize=12)
    if save_png:
        fig.savefig(save_png, dpi=150)
        print("[SAVED]", save_png)
        plt.close(fig)
    else:
        plt.show()

def plot_timeseries(pred4d, tgt4d, mask3d, subj_label, save_png=None):
    """画脑内均值随时间变化曲线（pred vs GT）"""
    Pm = pred4d[mask3d].reshape(-1, pred4d.shape[-1]).mean(axis=0)
    Tm = tgt4d[mask3d].reshape(-1, tgt4d.shape[-1]).mean(axis=0)
    fig = plt.figure(figsize=(6,3))
    plt.plot(Pm, label="Pred (brain mean)")
    plt.plot(Tm, label="GT (brain mean)")
    plt.xlabel("t")
    plt.ylabel("signal (mean over brain)")
    plt.title(f"{subj_label}  |  Δmean={float(np.mean(np.abs(Pm-Tm))):.4g}")
    plt.legend()
    if save_png:
        fig.savefig(save_png, dpi=150)
        print("[SAVED]", save_png)
        plt.close(fig)
    else:
        plt.show()

# ---------------- Load ----------------
subjects = np.array(load_gz_pickle(FILES["subjects"]))
pred     = np.asarray(load_gz_pickle(FILES["pred"]))  # [N, V*T]
tgt      = np.asarray(load_gz_pickle(FILES["tgt"]))   # [N, V*T]
N, flat_len = pred.shape
T_HZ = infer_T_horizon(flat_len, spatial=(96,96,96))
print(f"[INFO] dir={PRED_DIR}  tag={USE_TAG}  N={N}, flat_len={flat_len}, T_horizon={T_HZ}")

# ---------------- Score all & pick best/median/worst ----------------
scores = []
for i in range(N):
    p4 = to_4d(pred[i], T=T_HZ)
    g4 = to_4d(tgt[i],  T=T_HZ)
    msk = brain_mask_from_target(g4, thr=0.0)
    m   = mse_masked(p4, g4, msk)
    r   = r_time_mean(p4, g4, msk)
    scores.append((i, m, r))

scores = sorted(scores, key=lambda x: x[1])  # by MSE asc
idx_best, idx_med, idx_worst = scores[0][0], scores[len(scores)//2][0], scores[-1][0]
print("BEST  ", subjects[idx_best],  "MSE=", scores[0][1],  "r_time=", scores[0][2])
print("MEDIAN", subjects[idx_med],   "MSE=", scores[len(scores)//2][1], "r_time=", scores[len(scores)//2][2])
print("WORST ", subjects[idx_worst], "MSE=", scores[-1][1], "r_time=", scores[-1][2])

# ---------------- Plot chosen subjects ----------------
os.makedirs(os.path.join(PRED_DIR, "figs"), exist_ok=True)
for tag, idx in [("best", idx_best), ("median", idx_med), ("worst", idx_worst)]:
    p4 = to_4d(pred[idx], T=T_HZ)
    g4 = to_4d(tgt[idx],  T=T_HZ)
    msk = brain_mask_from_target(g4)
    plot_subject_grid(p4, g4, msk, subj_label=f"{subjects[idx]} ({tag})",
                      save_png=os.path.join(PRED_DIR, "figs", f"{tag}_{subjects[idx]}.png"))
    plot_timeseries(p4, g4, msk, subj_label=f"{subjects[idx]} ({tag})",
                    save_png=os.path.join(PRED_DIR, "figs", f"{tag}_{subjects[idx]}_ts.png"))
```

```python
# === Plot rfMRI_next signals (global mean, voxel, optional ROI) ===
import os, gzip, pickle, numpy as np, matplotlib.pyplot as plt

PRED_DIR = "./output/ds000030_rsfMRI_alldata_6/predictions/56an1qq8"   # <<< 修改为你的路径
USE_TAG  = "best"                     # "best" 或 "last"
SPATIAL  = (96,96,96)

# ---------- IO helpers ----------
def load_gz_pickle(path):
    with gzip.open(path, "rb") as f:
        return pickle.load(f)

subjects = np.array(load_gz_pickle(os.path.join(PRED_DIR, f"test_subjects_{USE_TAG}.pkl.gz")))
pred     = np.asarray(load_gz_pickle(os.path.join(PRED_DIR, f"predicted_map_{USE_TAG}.pkl.gz")))  # [N, V*T]
tgt      = np.asarray(load_gz_pickle(os.path.join(PRED_DIR, f"target_map_{USE_TAG}.pkl.gz")))     # [N, V*T]

V = np.prod(SPATIAL)
flat_len = pred.shape[1]
assert flat_len % V == 0, f"flat_len={flat_len} 不能整除 V={V}"
T = flat_len // V
print(f"[INFO] N={pred.shape[0]}, V={V}, T={T}")

def to_4d(x_flat):  # [V*T] -> [96,96,96,T]
    return x_flat.reshape(SPATIAL + (T,))

def brain_mask_from_target(t4d, thr=0.0):
    return (np.abs(t4d) > thr).any(axis=-1)   # 任一时刻非零的体素视为脑内

# ---------- 选择要看的 subject ----------
# 例：第 0 个；也可用 subject 名查找：idx = np.where(subjects == "sub-10280")[0][0]
idx = 0
print("Subject:", subjects[idx])

p4 = to_4d(pred[idx])
g4 = to_4d(tgt[idx])
mask3d = brain_mask_from_target(g4)

# ---------- 1) 全脑均值的时间序列（Global Signal） ----------
p_mean_t = p4[mask3d].reshape(-1, T).mean(axis=0)
g_mean_t = g4[mask3d].reshape(-1, T).mean(axis=0)

plt.figure(figsize=(6,3))
plt.plot(g_mean_t, label="GT brain-mean")
plt.plot(p_mean_t, label="Pred brain-mean", linestyle="--")
plt.xlabel("time (frames)")
plt.ylabel("signal (mean over brain)")
plt.title(f"{subjects[idx]} | Δmean={np.mean(np.abs(p_mean_t-g_mean_t)):.4g}")
plt.legend(); plt.show()

# ---------- 2) 指定体素的时间序列 ----------
# 方式A：给出体素索引 (x,y,z)
#x,y,z = 48,48,48   # <<< 改成你关心的坐标
x,y,z = 20,20,20 
p_vox = p4[x,y,z,:]
g_vox = g4[x,y,z,:]

print("GT voxel values:", g4[20,20,20,:])
print("GT brain mean across time:", g_mean_t)


plt.figure(figsize=(6,3))
plt.plot(g_vox, label=f"GT voxel({x},{y},{z})")
plt.plot(p_vox, label="Pred voxel", linestyle="--")
plt.xlabel("time (frames)"); plt.ylabel("signal")
plt.title(f"voxel @ ({x},{y},{z})"); plt.legend(); plt.show()

# ---------- 3) 可选：给一个 ROI mask，并画 ROI 均值曲线 ----------
# 如果你有 ROI 的三维布尔掩膜 (96,96,96)，命名为 roi_mask3d：
roi_mask3d = None  # <<< 放入你的 ROI 掩膜 (np.bool_ array)；没有就保持 None

if roi_mask3d is not None:
    assert roi_mask3d.shape == SPATIAL
    p_roi = p4[roi_mask3d].reshape(-1, T).mean(axis=0)
    g_roi = g4[roi_mask3d].reshape(-1, T).mean(axis=0)
    plt.figure(figsize=(6,3))
    plt.plot(g_roi, label="GT ROI-mean")
    plt.plot(p_roi, label="Pred ROI-mean", linestyle="--")
    plt.xlabel("time (frames)"); plt.ylabel("signal")
    plt.title("ROI mean time series"); plt.legend(); plt.show()

# ---------- 4) 简单的相关性指标（时间相关的平均） ----------
def r_time_mean(a4d, b4d, mask3d):
    A = a4d[mask3d].reshape(-1, a4d.shape[-1])
    B = b4d[mask3d].reshape(-1, b4d.shape[-1])
    Az = (A - A.mean(-1, keepdims=True)) / (A.std(-1, keepdims=True) + 1e-6)
    Bz = (B - B.mean(-1, keepdims=True)) / (B.std(-1, keepdims=True) + 1e-6)
    return float((Az*Bz).mean())

print("r_time_mean (brain mask):", r_time_mean(p4, g4, mask3d))
```

```python
# ==== rfMRI_next 直观可视化（含诊断）====
import os, gzip, pickle, numpy as np, matplotlib.pyplot as plt

# 1) 配置
PRED_DIR = "./output/ds000030_rsfMRI_alldata_6/predictions/56an1qq8"  # <<< 改成你的目录
USE_TAG  = "best"                    # "best" 或 "last"
SPATIAL  = (96,96,96)

# 2) 读取
def load_gz_pickle(path):
    with gzip.open(path, "rb") as f:
        return pickle.load(f)

subjects = np.array(load_gz_pickle(os.path.join(PRED_DIR, f"test_subjects_{USE_TAG}.pkl.gz")))
pred     = np.asarray(load_gz_pickle(os.path.join(PRED_DIR, f"predicted_map_{USE_TAG}.pkl.gz")))  # [N, V*T]
tgt      = np.asarray(load_gz_pickle(os.path.join(PRED_DIR, f"target_map_{USE_TAG}.pkl.gz")))     # [N, V*T]

V = np.prod(SPATIAL)
flat_len = pred.shape[1]
assert flat_len % V == 0, f"flat_len={flat_len} 不能整除 V={V}"
T = flat_len // V
print(f"[INFO] dir={PRED_DIR} tag={USE_TAG}  N={pred.shape[0]}, V={V}, T={T}")

def to_4d(x_flat):  # [V*T] -> [96,96,96,T]
    return np.asarray(x_flat).reshape(SPATIAL + (T,))

# 3) 选一个 subject
subj_idx = 0  # <<< 换成你想看的索引或用 np.where(subjects=="sub-xxxx")[0][0]
p4 = to_4d(pred[subj_idx])
g4 = to_4d(tgt[subj_idx])

# 4) 脑掩膜（在 GT 任一时间非零/非极小）
mask3d = (np.abs(g4) > 0).any(axis=-1)
nz = int(mask3d.sum())
print(f"[MASK] 非零体素数：{nz} / {V}  ({nz/V*100:.2f}%)")

# 5) 数值体检：帧内/帧间
def frame_stats(vol4d, mask):
    # 每帧（仅脑内）均值/标准差
    fm = []; fs = []
    for t in range(vol4d.shape[-1]):
        x = vol4d[...,t][mask]
        fm.append(float(x.mean()))
        fs.append(float(x.std()))
    return np.array(fm), np.array(fs)

pm, ps = frame_stats(p4, mask3d)
gm, gs = frame_stats(g4, mask3d)

print("[GT]  mean(min/median/max):", gm.min(), np.median(gm), gm.max())
print("[GT]   std(min/median/max):", gs.min(), np.median(gs), gs.max())
print("[Pred] mean(min/median/max):", pm.min(), np.median(pm), pm.max())
print("[Pred]  std(min/median/max):", ps.min(), np.median(ps), ps.max())

plt.figure(figsize=(7,3))
plt.plot(gm, label="GT mean"); plt.plot(pm, "--", label="Pred mean")
plt.title(f"{subjects[subj_idx]} | 全脑均值")
plt.xlabel("t"); plt.legend(); plt.show()

plt.figure(figsize=(7,3))
plt.plot(gs, label="GT std"); plt.plot(ps, "--", label="Pred std")
plt.title("全脑标准差（帧内，对比度大小）")
plt.xlabel("t"); plt.legend(); plt.show()

# 6) 空间 z-score（每帧在脑内做 z-score） -> 可把静态偏置去掉，更“有对比度”
def zscore_per_frame(vol4d, mask):
    out = np.empty_like(vol4d, dtype=np.float32)
    for t in range(vol4d.shape[-1]):
        x = vol4d[...,t]
        m = x[mask].mean(); s = x[mask].std() + 1e-6
        out[...,t] = (x - m) / s
    return out

p4z = zscore_per_frame(p4, mask3d)
g4z = zscore_per_frame(g4, mask3d)
diffz = p4z - g4z

# 7) 选择“信息量最大”的 z 切片（脑内体素最多）
z_counts = mask3d.sum(axis=(0,1))
z_idx = int(np.argmax(z_counts))

# 8) 显示 Pred/GT/DIFF（用 z-score 后的图）
def robust_limits(a, q=0.5):
    lo, hi = np.percentile(a, [q, 100-q])
    if lo == hi: hi = lo + 1e-6
    return float(lo), float(hi)

Tn = p4z.shape[-1]
fig, axes = plt.subplots(3, Tn, figsize=(2.4*Tn, 7), constrained_layout=True)
vmin, vmax = robust_limits(np.concatenate([p4z[:,:,z_idx,:].ravel(), g4z[:,:,z_idx,:].ravel()]))
dmax = float(np.max(np.abs(diffz[:,:,z_idx,:])))
dmin = -dmax

for t in range(Tn):
    for r,(title,vol) in enumerate([("Pred(z)", p4z), ("GT(z)", g4z), ("Diff(z)", diffz)]):
        ax = axes[r,t]
        img = vol[:,:,z_idx,t].T
        if title=="Diff(z)":
            ax.imshow(img, origin="lower", cmap="bwr", vmin=dmin, vmax=-dmin)
        else:
            ax.imshow(img, origin="lower", vmin=vmin, vmax=vmax)
        ax.set_axis_off()
        if r==0: ax.set_title(f"t={t+1}")
    axes[0,0].set_ylabel("Pred(z)", rotation=0, labelpad=20)
    axes[1,0].set_ylabel("GT(z)", rotation=0, labelpad=20)
    axes[2,0].set_ylabel("Diff(z)", rotation=0, labelpad=20)
fig.suptitle(f"Subject={subjects[subj_idx]} | z={z_idx}", fontsize=12)
plt.show()

# 9) voxelwise 相关图（时间相关；仅在非零 std 的体素内）
def voxelwise_corrmap(p4d, g4d, mask):
    V = np.prod(p4d.shape[:-1])
    P = p4d.reshape(V, -1); G = g4d.reshape(V, -1)
    mask_flat = mask.reshape(-1)
    valid = mask_flat.copy()
    # 去除时间方差≈0 的体素，避免 0/0
    valid[mask_flat] &= (G[mask_flat].std(1) > 1e-6)
    corr = np.zeros(V, dtype=np.float32); corr[:] = np.nan
    Pz = (P[valid]-P[valid].mean(1,keepdims=True))/(P[valid].std(1,keepdims=True)+1e-6)
    Gz = (G[valid]-G[valid].mean(1,keepdims=True))/(G[valid].std(1,keepdims=True)+1e-6)
    corr_valid = (Pz*Gz).mean(1)
    corr[valid] = corr_valid
    return corr.reshape(p4d.shape[:-1])

corrmap = voxelwise_corrmap(p4, g4, mask3d)
plt.figure(figsize=(5,4))
plt.imshow(corrmap[:,:,z_idx].T, origin="lower", cmap="bwr", vmin=-1, vmax=1)
plt.colorbar(label="Pred vs GT corr")
plt.title(f"Voxelwise corr (z={z_idx})  subj={subjects[subj_idx]}")
plt.show()

# 10) 选“方差最大的 K 个体素”画时序（最直观）
K = 10
g_flat = g4[mask3d].reshape(-1, T)
p_flat = p4[mask3d].reshape(-1, T)
top_idx = np.argsort(g_flat.std(1))[-K:]

plt.figure(figsize=(7,4))
for i in top_idx:
    plt.plot(g_flat[i], alpha=0.6)
plt.title("GT：方差 Top-K 体素的时间序列"); plt.xlabel("t")
plt.show()

plt.figure(figsize=(7,4))
for i in top_idx:
    plt.plot(p_flat[i], alpha=0.6)
plt.title("Pred：对应体素的时间序列"); plt.xlabel("t")
plt.show()
```
