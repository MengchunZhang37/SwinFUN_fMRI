```python
# dump_curves.py (fixed)
import os, glob, csv, re
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator

PROJECT_NAME = "ds000030_experiment_6"   # 改成你刚才实际打印出来的项目名
log_root = os.path.join("output", PROJECT_NAME, "lightning_logs")
versions = sorted(glob.glob(os.path.join(log_root, "version_*")), key=os.path.getmtime)
if not versions:
    raise SystemExit(f"No versions found under {log_root}")
log_dir = versions[-1]
print("Using log_dir:", log_dir)

ea = EventAccumulator(log_dir)
ea.Reload()
scalar_tags = set(ea.Tags().get("scalars", []))

def pick_tag(base):
    """在所有 scalar 里找以 base 开头的真实 tag，优先选 /dataloader_idx_0"""
    cands = [t for t in scalar_tags if t == base or t.startswith(base + "/")]
    if not cands:
        return None
    # 优先 dataloader_idx_0
    cands.sort(key=lambda x: 0 if x.endswith("/dataloader_idx_0") else 1)
    return cands[0]

def all_k(mode):
    """收集现有 tag 里出现过的 k（支持 ...@5/dataloader_idx_0 这种样式）"""
    ks = set()
    pat = re.compile(rf"^{mode}_mse@(\d+)(?:/.*)?$")
    for t in scalar_tags:
        m = pat.match(t)
        if m:
            ks.add(int(m.group(1)))
    return sorted(ks)

def dump_mode(mode):
    ks = all_k(mode)
    if ks:
        print(f"\n== {mode} k-step curves (last epoch) ==")
        print("k,mse_last,r_last")
        for k in ks:
            mse_tag = pick_tag(f"{mode}_mse@{k}")
            r_tag   = pick_tag(f"{mode}_r@{k}")
            mse = ea.Scalars(mse_tag)[-1].value if mse_tag else float("nan")
            r   = ea.Scalars(r_tag)[-1].value   if r_tag   else float("nan")
            print(f"{k},{mse:.6f},{r:.6f}")

        os.makedirs(os.path.join(log_dir, "exports"), exist_ok=True)
        with open(os.path.join(log_dir, "exports", f"{mode}_k_curves.csv"), "w", newline="") as f:
            w = csv.writer(f); w.writerow(["tag","step","value"])
            for k in ks:
                mse_tag = pick_tag(f"{mode}_mse@{k}")
                r_tag   = pick_tag(f"{mode}_r@{k}")
                if mse_tag:
                    for ev in ea.Scalars(mse_tag): w.writerow([mse_tag, ev.step, ev.value])
                if r_tag:
                    for ev in ea.Scalars(r_tag):   w.writerow([r_tag,   ev.step, ev.value])

    base_names = [
        f"{mode}_mse_baseline_persist", f"{mode}_mae_baseline_persist", f"{mode}_r_time_baseline_persist",
        f"{mode}_mse_baseline_ctxmean", f"{mode}_mae_baseline_ctxmean", f"{mode}_r_time_baseline_ctxmean",
    ]
    present = [(name, pick_tag(name)) for name in base_names]
    present = [(name, tag) for name, tag in present if tag is not None]
    if present:
        print(f"\n== {mode} baselines (last epoch) ==")
        for name, tag in present:
            print(f"{name}: {ea.Scalars(tag)[-1].value:.6f}")
        with open(os.path.join(log_dir, "exports", f"{mode}_baselines.csv"), "w", newline="") as f:
            w = csv.writer(f); w.writerow(["tag","step","value"])
            for name, tag in present:
                for ev in ea.Scalars(tag):
                    w.writerow([name, ev.step, ev.value])

dump_mode("valid")
dump_mode("test")
```

```python
log_dir = "output/ds000030_experiment_6/lightning_logs/version_0"  # 改成你的 version_x

import os, re
from collections import OrderedDict
from tensorboard.backend.event_processing import event_accumulator
import matplotlib.pyplot as plt

os.makedirs(os.path.join(log_dir, "plots"), exist_ok=True)

ea = event_accumulator.EventAccumulator(log_dir, size_guidance={"scalars": 0})
ea.Reload()
all_tags = ea.Tags().get("scalars", [])
base_tags = {t.split("/")[0] for t in all_tags}  # 去掉 /dataloader_idx_0 后缀

def _pick_physical_tag(base):
    """同一指标可能有多个后缀，优先选 dataloader_idx_0，其次选无后缀的最短名字"""
    cands = [t for t in all_tags if t.split("/")[0] == base]
    if not cands: 
        return None
    for prefer in ["dataloader_idx_0", "dataloader_idx_1"]:
        for t in cands:
            if prefer in t:
                return t
    return sorted(cands, key=len)[0]

def get_series(base):
    """返回 [(step, value), ...]（去重后按 step 的最后一次值）"""
    tag = _pick_physical_tag(base)
    if tag is None:
        return []
    es = ea.Scalars(tag)
    od = OrderedDict()
    for ev in es:
        od[ev.step] = ev.value
    items = list(od.items())
    items.sort(key=lambda x: x[0])
    return items

# ---------- A. k-step：最后一个 epoch 的 MSE/r 随 k 的曲线 ----------
ks = sorted({
    int(re.split(r"@", b)[1])
    for b in base_tags
    if b.startswith("valid_mse@")
})
# 找到最后一个 epoch（以 valid_mse@k 中出现的最大 step 为准）
steps_all = []
for k in ks:
    steps_all += [s for s, _ in get_series(f"valid_mse@{k}")]
step_last = max(steps_all) if steps_all else None

if step_last is not None and ks:
    mse_last = [dict(get_series(f"valid_mse@{k}")).get(step_last, None) for k in ks]
    r_last   = [dict(get_series(f"valid_r@{k}")).get(step_last, None)   for k in ks]

    # MSE vs k
    plt.figure()
    plt.plot(ks, mse_last, marker="o")
    plt.xlabel("k"); plt.ylabel("valid MSE (last epoch)"); plt.title("k-step MSE @ last epoch (valid)")
    plt.savefig(os.path.join(log_dir, "plots", "valid_mse_vs_k_last.png"), bbox_inches="tight")
    plt.show()

    # r vs k
    plt.figure()
    plt.plot(ks, r_last, marker="o")
    plt.xlabel("k"); plt.ylabel("valid r (last epoch)"); plt.title("k-step Pearson r @ last epoch (valid)")
    plt.savefig(os.path.join(log_dir, "plots", "valid_r_vs_k_last.png"), bbox_inches="tight")
    plt.show()

# ---------- B. k-step：随 epoch 的曲线（每个 k 一张图，避免太挤） ----------
for k in ks[:min(10, len(ks))]:   # 只画前10 个 k，按需增减
    series_k = get_series(f"valid_mse@{k}")
    if not series_k: 
        continue
    steps, vals = zip(*series_k)
    plt.figure()
    plt.plot(steps, vals, marker="o")
    plt.xlabel("epoch"); plt.ylabel(f"valid mse@{k}"); plt.title(f"valid mse@{k} over epochs")
    fname = f"valid_mse_at_k{k}_over_epochs.png"
    plt.savefig(os.path.join(log_dir, "plots", fname), bbox_inches="tight")
    plt.show()

    series_k = get_series(f"valid_r@{k}")
    if series_k:
        steps, vals = zip(*series_k)
        plt.figure()
        plt.plot(steps, vals, marker="o")
        plt.xlabel("epoch"); plt.ylabel(f"valid r@{k}"); plt.title(f"valid r@{k} over epochs")
        fname = f"valid_r_at_k{k}_over_epochs.png"
        plt.savefig(os.path.join(log_dir, "plots", fname), bbox_inches="tight")
        plt.show()

# ---------- C. 基线 & 模型：随 epoch 的 MSE 曲线（valid/test 各一张） ----------
def plot_baselines(mode):
    plt.figure()
    plotted = False
    for name in ["mse_baseline_persist", "mse_baseline_ctxmean"]:
        s = get_series(f"{mode}_{name}")
        if s:
            steps, vals = zip(*s)
            plt.plot(steps, vals, marker="o", label=name)
            plotted = True
    s_model = get_series(f"{mode}_mse")
    if s_model:
        steps, vals = zip(*s_model)
        # 只是一条虚线区分一下（不指定颜色）
        plt.plot(steps, vals, marker="o", linestyle="--", label="model_mse")
        plotted = True
    if plotted:
        plt.xlabel("epoch"); plt.ylabel(f"{mode} MSE"); plt.title(f"{mode}: model vs baselines")
        plt.legend()
        plt.savefig(os.path.join(log_dir, "plots", f"{mode}_model_vs_baselines.png"), bbox_inches="tight")
        plt.show()

plot_baselines("valid")
plot_baselines("test")

print("✅ 已保存到：", os.path.join(log_dir, "plots"))
```
